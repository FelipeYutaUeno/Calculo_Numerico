```python
import math
import numpy as np
import matplotlib.pyplot as plt
plt.style.use("seaborn-v0_8-whitegrid")
```

```python
x1 = 3.14159265
x2 = 98.76543
x3 = 0.00498765

n = [0, 1, 2, 3, 4]
```

```python
x1arre = [round(x1, i) for i in n]
x2arre = [round(x2, i) for i in n]
x3arre = [round(x3, i) for i in n]

print(x1arre)
print(x2arre)
print(x3arre)
```

```python
x1trunc = [(math.trunc((10**i)*x1))/(10**i) for i in n]
x2trunc = [(math.trunc((10**i)*x2))/(10**i) for i in n]
x3trunc = [(math.trunc((10**i)*x3))/(10**i) for i in n]

print(x1trunc)
print(x2trunc)
print(x3trunc)
```

```python
def Eabs(x, xn):
  return abs(x - xn)

def Erel(x, xn):
  return (abs(x - xn)/abs(x))

def Eper(x, xn):
  return 100*Erel(x, xn)
```

```python
ErrAbsX1 = [Eabs(x1, xn) for xn in x1arre]
print(f"Erros absolutos x1: {ErrAbsX1}")
ErrRelX1 = [Erel(x1, xn) for xn in x1arre]
print(f"Erros Relativos x1: {ErrRelX1}")
ErrPerX1 = [Eper(x1, xn) for xn in x1arre]
print(f"Erros Percentagens x1: {ErrPerX1}")
```

```python
ErrAbsX2 = [Eabs(x2, xn) for xn in x2arre]
print(f"Erros absolutos x2: {ErrAbsX2}")
ErrRelX2 = [Erel(x2, xn) for xn in x2arre]
print(f"Erros Relativos x2: {ErrRelX2}")
ErrPerX2 = [Eper(x2, xn) for xn in x2arre]
print(f"Erros Percentagens x2: {ErrPerX2}")
```

```python
ErrAbsX3 = [Eabs(x3, xn) for xn in x3arre]
print(f"Erros absolutos x3: {ErrAbsX3}")
ErrRelX3 = [Erel(x3, xn) for xn in x3arre]
print(f"Erros Relativos x3: {ErrRelX3}")
ErrPerX3 = [Eper(x3, xn) for xn in x3arre]
print(f"Erros Percentagens x3: {ErrPerX3}")
```

```python
ErrAbsX1Trunc = [Eabs(x1, xn) for xn in x1trunc]
print(f"Erros absolutos x1Trunc: {ErrAbsX1Trunc}")
ErrRelX1Trunc = [Erel(x1, xn) for xn in x1trunc]
print(f"Erros Relativos x1 TRrunc: {ErrRelX1Trunc}")
ErrPerX1Trunc = [Eper(x1, xn) for xn in x1trunc]
print(f"Erros Percentagens x1 Trunc: {ErrPerX1Trunc}")
```

```python
ErrAbsX2Trunc = [Eabs(x2, xn) for xn in x2trunc]
print(f"Erros absolutos x2 Trunc: {ErrAbsX2Trunc}")
ErrRelX2Trunc = [Erel(x2, xn) for xn in x2trunc]
print(f"Erros Relativos x2 Trunc: {ErrRelX2Trunc}")
ErrPerX2Trunc = [Eper(x2, xn) for xn in x2trunc]
print(f"Erros Percentagens x2 Trunc: {ErrPerX2Trunc}")
```

```python
ErrAbsX3Trunc = [Eabs(x3, xn) for xn in x3trunc]
print(f"Erros absolutos x3 Trunc: {ErrAbsX3Trunc}")
ErrRelX3Trunc = [Erel(x3, xn) for xn in x3trunc]
print(f"Erros Relativos x3 Trunc: {ErrRelX3Trunc}")
ErrPerX3Trunc = [Eper(x3, xn) for xn in x3trunc]
print(f"Erros Percentagens x3 Trunc: {ErrPerX3Trunc}")
```

# Parte B: Medidas de pressão arterial
121,7; 120,4; 122,1; 119,8; 121,2; 120,9; 122,4; 121,0; 120,2; 121,5.

```python
medidasOriginais = [121.7, 120.4, 122.1, 119.8, 121.2, 120.9, 122.4, 121.0, 120.2, 121.5]

medidasArredondadas = [round(medida, 0) for medida in medidasOriginais]

medidasTruncadas = [math.trunc(medida) for medida in medidasOriginais]

print(medidasOriginais)
print(medidasArredondadas)
print(medidasTruncadas)

media = np.mean(medidasOriginais)

print(f"\nMédia: {media}")
```

```python
def calculate_stats(data, name):
    mean_val = np.mean(data)
    min_val = np.min(data)
    max_val = np.max(data)
    amplitude = max_val - min_val
    std_dev = np.std(data, ddof=1)
    print(f"\n--- Estatísticas para {name} ---")
    print(f"Média: {mean_val:.4f}")
    print(f"Mínimo: {min_val:.4f}")
    print(f"Máximo: {max_val:.4f}")
    print(f"Amplitude: {amplitude:.4f}")
    print(f"Desvio-padrão amostral: {std_dev:.4f}")
    return mean_val

mean_original = calculate_stats(medidasOriginais, "Dados Originais")
mean_arredondadas = calculate_stats(medidasArredondadas, "Dados Arredondados")
mean_truncadas = calculate_stats(medidasTruncadas, "Dados Truncados")

print("\n--- Erros da Média Arredondada em relação à Média Original ---")
print(f"Média de Referência (Original): {mean_original:.4f}")
print(f"Média Arredondada: {mean_arredondadas:.4f}")
print(f"Erro Absoluto da Média Arredondada: {Eabs(mean_original, mean_arredondadas):.6f}")
print(f"Erro Relativo da Média Arredondada: {Erel(mean_original, mean_arredondadas):.6f}")
print(f"Erro Percentual da Média Arredondada: {Eper(mean_original, mean_arredondadas):.6f}%")

print("\n--- Erros da Média Truncada em relação à Média Original ---")
print(f"Média de Referência (Original): {mean_original:.4f}")
print(f"Média Truncada: {mean_truncadas:.4f}")
print(f"Erro Absoluto da Média Truncada: {Eabs(mean_original, mean_truncadas):.6f}")
print(f"Erro Relativo da Média Truncada: {Erel(mean_original, mean_truncadas):.6f}")
print(f"Erro Percentual da Média Truncada: {Eper(mean_original, mean_truncadas):.6f}%")
```

# Parte C: Vazão em um tubo de pequeno diâmetro (Hagen-Poiseuille)

Q = pi * r^4 * dp / (8 * mu * L), com r = 0,80 mm, L = 0,20 m, mu = 3,5x10^-3 Pa.s, dp = 1200 Pa.

```python

def vazaoHP(r, dp, mu, L):
    return (math.pi * r**4 * dp) / (8 * mu * L)

rC = 0.80e-3   # m
LC = 0.20      # m
muC = 3.5e-3   # Pa.s
dpC = 1200.0   # Pa

QrefC = vazaoHP(rC, dpC, muC, LC)
print(f"Q de referência = {QrefC:.6e} m3/s")
```

```python

# C.1 - Precisao das variaveis: r, mu e dp aproximados (isolados e combinados)
rC_mm = rC * 1e3
rC_round = round(rC_mm, 1) / 1e3
rC_trunc = (math.trunc(rC_mm * 10)) / 10 / 1e3

muC_round = round(muC, 3)
muC_trunc = (math.trunc(muC * 1000)) / 1000

dpC_round = round(dpC, -2)
dpC_trunc = math.trunc(dpC / 100) * 100

cenariosC = {
    "1. valores fornecidos (referencia)": (rC, dpC, muC),
    "2a. r arredondado (1 casa mm)":       (rC_round, dpC, muC),
    "2b. r truncado (1 casa mm)":          (rC_trunc, dpC, muC),
    "3a. mu arredondado (3 casas)":        (rC, dpC, muC_round),
    "3b. mu truncado (3 casas)":           (rC, dpC, muC_trunc),
    "4a. dp arredondado (centena)":        (rC, dpC_round, muC),
    "4b. dp truncado (centena)":           (rC, dpC_trunc, muC),
    "5a. todas arredondadas":              (rC_round, dpC_round, muC_round),
    "5b. todas truncadas":                 (rC_trunc, dpC_trunc, muC_trunc),
}

linhasC1 = []
for nome, (r_, dp_, mu_) in cenariosC.items():
    Q = vazaoHP(r_, dp_, mu_, LC)
    linhasC1.append({
        "cenario": nome, "r_m": r_, "dp_Pa": dp_, "mu_Pa_s": mu_, "Q_m3_s": Q,
        "Erro Absoluto": Eabs(QrefC, Q), "Erro Relativo": Erel(QrefC, Q), "Erro Percentual": Eper(QrefC, Q),
    })

df_C1 = pd.DataFrame(linhasC1)
print(df_C1.to_string(index=False))
```

```python

# C.2 - Sensibilidade: varia r de 0,70 a 0,90 mm
rsC_mm = [0.70 + 0.01*i for i in range(21)]
rsC_m = [r/1000 for r in rsC_mm]

Qreferencia = [vazaoHP(r, dpC, muC, LC) for r in rsC_m]
rs_round_m = [round(r, 1)/1000 for r in rsC_mm]
rs_trunc_m = [(math.trunc(r*10))/10/1000 for r in rsC_mm]
Qround = [vazaoHP(r, dpC, muC, LC) for r in rs_round_m]
Qtrunc = [vazaoHP(r, dpC, muC, LC) for r in rs_trunc_m]

EaRound = [Eabs(qr, qa) for qr, qa in zip(Qreferencia, Qround)]
EaTrunc = [Eabs(qr, qa) for qr, qa in zip(Qreferencia, Qtrunc)]
ErRound = [Erel(qr, qa) for qr, qa in zip(Qreferencia, Qround)]
ErTrunc = [Erel(qr, qa) for qr, qa in zip(Qreferencia, Qtrunc)]
EpRound = [Eper(qr, qa) for qr, qa in zip(Qreferencia, Qround)]
EpTrunc = [Eper(qr, qa) for qr, qa in zip(Qreferencia, Qtrunc)]

fig, axes = plt.subplots(2, 2, figsize=(12, 9))

axes[0,0].plot(rsC_mm, Qreferencia, 'o-', label='Q referencia')
axes[0,0].plot(rsC_mm, Qround, 's--', label='Q (r arredondado)')
axes[0,0].plot(rsC_mm, Qtrunc, '^:', label='Q (r truncado)')
axes[0,0].set_title('Q em funcao de r'); axes[0,0].set_xlabel('r (mm)'); axes[0,0].set_ylabel('Q (m3/s)'); axes[0,0].legend()

axes[0,1].plot(rsC_mm, EaRound, 's--', label='arredondado')
axes[0,1].plot(rsC_mm, EaTrunc, '^:', label='truncado')
axes[0,1].set_title('Erro absoluto de Q vs. r'); axes[0,1].set_xlabel('r (mm)'); axes[0,1].set_ylabel('Ea (m3/s)'); axes[0,1].legend()

axes[1,0].plot(rsC_mm, ErRound, 's--', label='arredondado')
axes[1,0].plot(rsC_mm, ErTrunc, '^:', label='truncado')
axes[1,0].set_title('Erro relativo de Q vs. r'); axes[1,0].set_xlabel('r (mm)'); axes[1,0].set_ylabel('Er'); axes[1,0].legend()

axes[1,1].plot(rsC_mm, EpRound, 's--', label='arredondado')
axes[1,1].plot(rsC_mm, EpTrunc, '^:', label='truncado')
axes[1,1].set_title('Erro percentual de Q vs. r'); axes[1,1].set_xlabel('r (mm)'); axes[1,1].set_ylabel('E%'); axes[1,1].legend()

fig.suptitle('C.2 - Sensibilidade de Q ao raio')
plt.tight_layout()
plt.show()
```

**Por que o erro no raio é amplificado?** Como Q é proporcional a r^4, um erro relativo delta no
raio se propaga aproximadamente como 4*delta na vazão (`DeltaQ/Q ~= 4 * Deltar/r`). O expoente 4
funciona como um "multiplicador" do erro relativo: pequenas variações geométricas no raio (por
exemplo, de arredondamento ou truncamento) produzem um efeito quatro vezes maior,
proporcionalmente, na vazão.

# Parte D: Escoamento em um nanocateter

r = 50 nm, L = 10 um, mu = 3,5x10^-3 Pa.s, dp = 5000 Pa.

```python

rD = 50e-9     # m
LD = 10e-6     # m
muD = 3.5e-3   # Pa.s
dpD = 5000.0   # Pa

QrefD = vazaoHP(rD, dpD, muD, LD)
print(f"Q de referencia (nanocateter) = {QrefD:.6e} m3/s")
```

```python

# D.1 - Erro geometrico: dr = 0.1 .. 5.0 nm
drsD_nm = [round(0.1*i, 1) for i in range(1, 51)]
drsD_m = [dr*1e-9 for dr in drsD_nm]

Qmenos = [vazaoHP(rD - dr, dpD, muD, LD) for dr in drsD_m]
Qmais  = [vazaoHP(rD + dr, dpD, muD, LD) for dr in drsD_m]

EaMax = [max(Eabs(QrefD, qm), Eabs(QrefD, qp)) for qm, qp in zip(Qmenos, Qmais)]
ErMax = [max(Erel(QrefD, qm), Erel(QrefD, qp)) for qm, qp in zip(Qmenos, Qmais)]
EpMax = [100*e for e in ErMax]

aproxLinear = [4*dr/rD for dr in drsD_m]

fig, axes = plt.subplots(2, 2, figsize=(12, 9))

axes[0,0].plot(drsD_nm, Qmenos, label='Q(r-dr)')
axes[0,0].axhline(QrefD, color='gray', linestyle='--', label='Q(r)')
axes[0,0].plot(drsD_nm, Qmais, label='Q(r+dr)')
axes[0,0].set_title('Q(r-), Q(r), Q(r+)'); axes[0,0].set_xlabel('dr (nm)'); axes[0,0].set_ylabel('Q (m3/s)'); axes[0,0].legend()

axes[0,1].plot(drsD_nm, EaMax, color='#C44E52')
axes[0,1].set_title('Erro absoluto maximo de Q'); axes[0,1].set_xlabel('dr (nm)'); axes[0,1].set_ylabel('Ea (m3/s)')

axes[1,0].plot(drsD_nm, ErMax, label='erro relativo real (max.)')
axes[1,0].plot(drsD_nm, aproxLinear, '--', label='aprox. linear 4*dr/r')
axes[1,0].set_title('Erro relativo maximo de Q'); axes[1,0].set_xlabel('dr (nm)'); axes[1,0].set_ylabel('Er'); axes[1,0].legend()

axes[1,1].plot(drsD_nm, EpMax, color='#55A868')
axes[1,1].set_title('Erro percentual maximo de Q'); axes[1,1].set_xlabel('dr (nm)'); axes[1,1].set_ylabel('E%')

fig.suptitle('D.1 - Erro geometrico no nanocateter')
plt.tight_layout()
plt.show()

print('Erro relativo maximo real vs. aproximacao linear (dr=5nm):')
print('real:', ErMax[-1], ' | linear:', aproxLinear[-1])
```

A aproximação linear `DeltaQ/Q ~= 4*Deltar/r` subestima o erro real conforme Δr cresce (ela é
válida apenas para Δr/r pequeno, sendo a expansão de 1a ordem de `(1+Deltar/r)^4`). Para Δr = 5 nm
(10% do raio de 50 nm), a diferença entre a estimativa linear e o erro real já é visível no gráfico.

```python

# D.2 - Precisao computacional: float32 vs float64
r32, dp32, mu32, L32 = np.float32(rD), np.float32(dpD), np.float32(muD), np.float32(LD)
r64, dp64, mu64, L64 = np.float64(rD), np.float64(dpD), np.float64(muD), np.float64(LD)

r4_32 = r32**4
r4_64 = r64**4
Q32 = vazaoHP(r32, dp32, mu32, L32)
Q64 = vazaoHP(r64, dp64, mu64, L64)

diffAbs = abs(float(Q32) - float(Q64))
diffRel = diffAbs / abs(float(Q64))

print(f"r^4 (float32): {r4_32}")
print(f"r^4 (float64): {r4_64}")
print(f"Q (float32): {Q32:.6e}")
print(f"Q (float64): {Q64:.6e}")
print(f"Diferenca absoluta Q32 vs Q64: {diffAbs:.3e} m3/s")
print(f"Diferenca relativa Q32 vs Q64: {diffRel:.3e}")
print(f"eps(float32) = {np.finfo(np.float32).eps:.3e}")
print(f"eps(float64) = {np.finfo(np.float64).eps:.3e}")
```

```python

# Comparacao delta1 (direto) vs delta2 (expandido) para Deltar/r decrescente - cancelamento catastrofico
drsTeste = [1e-6, 1e-7, 1e-8, 1e-9, 1e-10, 1e-11, 1e-12, 1e-13, 1e-14, 1e-15]

linhasD2 = []
for dr in drsTeste:
    delta1 = (vazaoHP(rD + dr, dpD, muD, LD) - QrefD) / QrefD
    delta2 = (1 + dr/rD)**4 - 1
    linhasD2.append({
        "dr_m": dr, "dr_sobre_r": dr/rD,
        "delta1_direto": delta1, "delta2_expandido": delta2,
        "diferenca_absoluta": abs(delta1 - delta2),
    })

df_D2 = pd.DataFrame(linhasD2)
print(df_D2.to_string(index=False))
```

Para Δr/r muito pequeno, `delta1` (calculado por subtração direta `Q(r+Δr) - Q(r)`, dividida por
Q(r)) sofre **cancelamento catastrófico**: os dois valores de Q ficam quase idênticos em ponto
flutuante, então a subtração perde algarismos significativos e o resultado passa a ser dominado
pelo erro de arredondamento de máquina (da ordem de `eps` ≈ 2,22x10^-16 para float64). Já
`delta2`, obtido pela expansão algébrica `(1+Δr/r)^4 - 1` (que evita subtrair dois números quase
iguais), permanece numericamente estável mesmo para Δr/r muito pequeno. Isso mostra por que, em
cálculo numérico, reescrever uma expressão algebricamente equivalente pode evitar perda de
precisão.

**D.3 - Fenômeno físico:** ao reduzir a escala geométrica de um tubo macroscópico (raio em
milímetros) para um nanocateter (raio em dezenas de nanômetros), a razão entre a área de
superfície e o volume do canal aumenta muito (superfície decai com r^2, volume decai com r^3,
então superfície/volume cresce quando r diminui). Isso faz com que efeitos de parede — atrito
viscoso, camada limite, interações moleculares entre o fluido e a superfície — passem a dominar o
escoamento, algo desprezível em tubos maiores. Combinado ao expoente 4 da lei de
Hagen-Poiseuille, isso explica por que pequenas variações no raio do nanocateter (poucos
nanômetros) podem alterar drasticamente a vazão: o resultado pode ser numericamente muito preciso
(muitas casas decimais corretas em float64) e, ainda assim, fisicamente inadequado, caso o
próprio modelo contínuo de Hagen-Poiseuille deixe de valer nessa escala. Precisão aritmética e
validade do modelo físico são conceitos diferentes.

## Conclusão geral

O arredondamento é sistematicamente mais preciso que o truncamento para qualquer número de casas
decimais, tanto nos exemplos numéricos isolados (Parte A) quanto nas médias de pressão arterial
(Parte B). Na propagação de erros em vazão (Partes C e D), o expoente 4 da lei de
Hagen-Poiseuille amplifica fortemente qualquer imprecisão no raio, efeito que se torna crítico em
escala nanométrica, onde poucos nanômetros de incerteza geram variações de dezenas de por cento
na vazão. A comparação float32/float64 e a análise de cancelamento catastrófico (delta1 vs.
delta2) mostram que a forma como uma expressão é calculada também afeta a precisão do resultado —
e que precisão numérica não garante, por si só, validade física do modelo em escalas onde efeitos
de superfície passam a dominar o escoamento.
